```python
# ============================================================
# Figure 3.1: Experimental Design Schematic
# ============================================================
#
# This notebook generates the experimental-design figure used in Chapter 3
# of the dissertation "Explainable AI for Credit Risk Assessment".
#
# The figure contains no analytical results. It summarises the structure of
# the study by linking the two datasets, four predictive models, explanation
# methods and statistical analyses used to address the three research questions.
#
# The figure is generated programmatically and saved to the local images
# directory used by the repository.
# ============================================================

import os
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# Save generated figures to the repository's local images directory
OUT_DIR = 'images'
os.makedirs(OUT_DIR, exist_ok=True)
```

```python
# ============================================================
# Figure Settings and Drawing Helpers
# ============================================================

# Muted palette so the figure remains legible in greyscale as well as colour
TEAL   = "#5B9C8F"   # Column headings
GREY   = "#4A4A4A"   # Body text
RED    = "#B5556A"   # Annotations and comparative context
LIGHT  = "#EDF3F1"   # Box fill
EDGE   = "#7A7A7A"   # Box borders
RQFILL = "#E8F0EE"   # Research question boxes

# LaTeX scales the figure to text width, so this canvas keeps labels legible
fig, ax = plt.subplots(figsize=(9.0, 4.58))

ax.set_xlim(0.05, 9.95)
ax.set_ylim(0, 5.6)
ax.axis("off")


def box(x, y, w, h, text, fc=LIGHT, ec=EDGE, fs=8.5, bold=False):
    """Draw a rounded labelled box."""
    ax.add_patch(
        FancyBboxPatch(
            (x, y),
            w,
            h,
            boxstyle="round,pad=0.06",
            fc=fc,
            ec=ec,
            lw=1.0,
            zorder=2
        )
    )

    ax.text(
        x + w / 2,
        y + h / 2,
        text,
        ha='center',
        va='center',
        fontsize=fs,
        color=GREY,
        zorder=3,
        linespacing=1.4,
        fontweight='bold' if bold else 'normal'
    )


def arrow(x1, y1, x2, y2):
    """Draw a directional arrow between components."""
    ax.add_patch(
        FancyArrowPatch(
            (x1, y1),
            (x2, y2),
            arrowstyle='-|>',
            mutation_scale=11,
            lw=1.0,
            color=EDGE,
            zorder=1
        )
    )
```

```python
# ============================================================
# Draw Experimental Design
# ============================================================

# Four columns represent the analytical workflow:
# datasets -> models -> explanation methods -> research questions

for x, label in [
    (0.9,  "Datasets"),
    (3.2,  "Models"),
    (5.78, "Explanation\nMethods"),
    (8.6,  "Research Questions\nand Tests")
]:
    ax.text(
        x,
        5.25,
        label,
        ha='center',
        va='center',
        fontsize=10.5,
        fontweight='bold',
        color=TEAL
    )


# ------------------------------------------------------------
# Column 1: Datasets
# ------------------------------------------------------------

# The datasets provide contrasting modelling settings.
# Their difference in sample size is important, but the dissertation does
# not treat sample size as a causal explanation for the observed results.

box(
    0.15,
    3.55,
    1.5,
    0.85,
    "Taiwan\n~30,000 Rows"
)

box(
    0.15,
    2.25,
    1.5,
    0.85,
    "German\n1,000 Rows"
)

ax.text(
    0.9,
    1.85,
    "Contrasting Data\nConditions",
    ha='center',
    va='center',
    fontsize=8,
    style='italic',
    color=RED
)


# ------------------------------------------------------------
# Column 2: Models
# ------------------------------------------------------------

# The models range from linear logistic regression to more flexible
# nonlinear approaches.

for label, y in [
    ("Logistic Regression", 4.30),
    ("XGBoost",             3.55),
    ("TensorFlow NN",       2.80),
    ("PyTorch NN",          2.05)
]:
    box(
        2.30,
        y,
        1.85,
        0.6,
        label
    )

ax.text(
    3.2,
    1.72,
    "Linear $\\rightarrow$ Nonlinear",
    ha='center',
    va='center',
    fontsize=8,
    style='italic',
    color=RED
)

# Bracket the two neural networks because the matched architecture supports
# the TensorFlow and PyTorch reproducibility comparison in RQ2.

ax.plot(
    [4.29, 4.41, 4.41, 4.29],
    [2.08, 2.08, 3.37, 3.37],
    color=RED,
    lw=1.1
)

ax.text(
    4.48,
    2.72,
    "Matched\nPair",
    ha='left',
    va='center',
    fontsize=7.5,
    color=RED
)


# ------------------------------------------------------------
# Column 3: Explanation Methods
# ------------------------------------------------------------

# The methods have different theoretical and computational foundations,
# motivating empirical comparison rather than assuming agreement.

for label, y in [
    ("SHAP",                 4.10),
    ("Integrated Gradients", 3.25),
    ("LIME",                 2.40)
]:
    box(
        4.85,
        y,
        1.85,
        0.6,
        label
    )

ax.text(
    5.78,
    2.08,
    "Differing Formal Bases",
    ha='center',
    va='center',
    fontsize=8,
    style='italic',
    color=RED
)


# ------------------------------------------------------------
# Column 4: Research Questions and Tests
# ------------------------------------------------------------

box(
    7.35,
    4.15,
    2.5,
    0.75,
    "RQ1  Concordance\nKendall's $W$, Spearman",
    fc=RQFILL
)

box(
    7.35,
    3.05,
    2.5,
    0.75,
    "RQ2  Reproducibility\nWilcoxon + Bonferroni",
    fc=RQFILL
)

box(
    7.35,
    1.95,
    2.5,
    0.75,
    "RQ3  Fairness\nFour Fifths + Bootstrap",
    fc=RQFILL
)


# ------------------------------------------------------------
# Arrows Between Columns
# ------------------------------------------------------------

# Datasets feed the predictive models
arrow(
    1.70,
    3.3,
    2.25,
    3.3
)

# Models feed the explanation methods
arrow(
    4.60,
    3.3,
    4.80,
    3.3
)

# Explanation analysis contributes to the three research questions
for y in (4.52, 3.42, 2.32):
    arrow(
        6.75,
        3.3,
        7.30,
        y
    )


# ------------------------------------------------------------
# Common Interpretive Context
# ------------------------------------------------------------

# Data availability and fitted model complexity are interpreted as plausible
# contributors to the cross-dataset patterns. The study does not establish
# either factor as causal.

box(
    0.15,
    0.45,
    9.7,
    0.72,
    "Possible common context: data availability and fitted model complexity "
    "may contribute to variation across the three analyses",
    fc="#F5E9EC",
    ec=RED,
    fs=8.5,
    bold=True
)

for x in (2.2, 5.0, 8.6):
    arrow(
        x,
        1.90 if x > 7 else 1.68,
        x,
        1.22
    )
```

```python
# ============================================================
# Save Figure
# ============================================================

# Use the same filename referenced by the dissertation
out_path = os.path.join(
    OUT_DIR,
    'fig_experimental_design.png'
)

plt.tight_layout()

fig.savefig(
    out_path,
    dpi=220,
    bbox_inches='tight',
    facecolor='white'
)

plt.show()

print("Written:", out_path)
```
